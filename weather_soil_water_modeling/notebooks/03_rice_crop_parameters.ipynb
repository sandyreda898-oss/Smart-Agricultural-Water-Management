```python
crop_calendar_clean = crop_calendar.copy()
season_clean = season.copy()

date_columns = [
    "season_start",
    "season_end",
    "stage_start_date",
    "stage_end_date"]

for col in date_columns:
    crop_calendar_clean[col] = pd.to_datetime(
        crop_calendar_clean[col],
        errors="coerce")

season_clean["start_date"] = pd.to_datetime(
    season_clean["start_date"],
    errors="coerce")

season_clean["end_date"] = pd.to_datetime(
    season_clean["end_date"],
    errors="coerce")

print("Invalid crop calendar dates:")
print(crop_calendar_clean[date_columns].isnull().sum())

print("\nInvalid season dates:")
print(season_clean[["start_date", "end_date"]].isnull().sum())
```

```python
print(
    "Calendar rows where start >= end:",
    (crop_calendar_clean["season_start"] >= crop_calendar_clean["season_end"]).sum())

print(
    "Season rows where start >= end:",
    (season_clean["start_date"] >= season_clean["end_date"]).sum())

print(
    "Calendar years:",
    crop_calendar_clean["season_year"].min(),
    "to",
    crop_calendar_clean["season_year"].max())

print(
    "Season years:",
    season_clean["season_year"].min(),
    "to",
    season_clean["season_year"].max())
```

```python
calendar_season_check = crop_calendar_clean.merge(
    season_clean,
    on="season_year",
    how="left",
    validate="one_to_one"
)

calendar_season_check[
    [
        "season_year",
        "season_start",
        "start_date",
        "season_end",
        "end_date"
    ]
]
```

```python
print(
    "Start-date mismatches:",
    (
        calendar_season_check["season_start"] !=
        calendar_season_check["start_date"]
    ).sum()
)

print(
    "End-date mismatches:",
    (
        calendar_season_check["season_end"] !=
        calendar_season_check["end_date"]
    ).sum()
)
```

```python
crop_water_clean = crop_water.copy()

crop_water_clean["parameter_scope"] = np.where(
    crop_water_clean["variety_id"].isna(),
    "GENERAL_RICE",
    "VARIETY_SPECIFIC"
)

crop_water_clean["is_range"] = (
    crop_water_clean["value_min"] !=
    crop_water_clean["value_max"]
)

crop_water_clean[
    [
        "parameter_id",
        "growth_stage",
        "parameter_name",
        "parameter_value",
        "kc",
        "value_min",
        "value_max",
        "unit",
        "parameter_scope",
        "is_range"
    ]
]
```

```python
print(
    "Invalid min/max ranges:",
    (
        crop_water_clean["value_min"] >
        crop_water_clean["value_max"]
    ).sum()
)
```

```python
kc_parameters = crop_water_clean[
    crop_water_clean["parameter_name"] == "Kc"
][
    [
        "growth_stage",
        "kc",
        "value_min",
        "value_max",
        "unit",
        "data_status"
    ]
]

kc_parameters
```

```python
stage_lengths = crop_water_clean[
    crop_water_clean["parameter_name"] == "stage_length"
][
    [
        "growth_stage",
        "parameter_value",
        "unit",
        "data_status"
    ]
]

stage_lengths
```

```python
total_reference_stage_days = stage_lengths[
    "parameter_value"
].sum()

print(
    "Total reference stage length:",
    total_reference_stage_days,
    "days"
)
```

```python
rice_variety_clean = rice_variety.merge(
    rice_variety_reference,
    on="variety_id",
    how="left",
    validate="one_to_one",
    suffixes=("_variety", "_reference")
)

print("Merged variety rows:", len(rice_variety_clean))

rice_variety_clean.head()
```

```python
rice_variety_clean[
    rice_variety_clean["source_url_reference"].isna()
][
    [
        "variety_id",
        "name_en",
        "data_status_variety"
    ]
]
```

```python
rice_variety_clean[
    rice_variety_clean["season_length_days"].isna()
][
    [
        "variety_id",
        "name_en",
        "season_length_days",
        "data_status_variety"
    ]
]
```

```python
rice_variety_clean["season_length_days"].describe()
```

```python
rice_variety_clean[
    rice_variety_clean["season_length_days"].notna()
][
    [
        "variety_id",
        "name_en",
        "season_length_days"
    ]
].sort_values("season_length_days")
```

```python
rice_variety_clean["season_length_available"] = (
    rice_variety_clean["season_length_days"].notna()
)

rice_variety_clean["reference_available"] = (
    rice_variety_clean["source_url_reference"].notna()
)

rice_variety_clean[
    [
        "variety_id",
        "name_en",
        "season_length_available",
        "reference_available"
    ]
]
```

```python
stage_reference = stage_lengths[
    ["growth_stage", "parameter_value", "unit", "data_status"]
].copy()

stage_reference = stage_reference.rename(
    columns={
        "parameter_value": "stage_length_days",
        "unit": "stage_length_unit",
        "data_status": "stage_length_status"
    }
)

stage_reference = stage_reference.merge(
    kc_parameters[
        [
            "growth_stage",
            "kc",
            "value_min",
            "value_max",
            "data_status"
        ]
    ],
    on="growth_stage",
    how="left"
)

stage_reference = stage_reference.rename(
    columns={
        "value_min": "kc_min",
        "value_max": "kc_max",
        "data_status": "kc_status"
    }
)

stage_reference
```

```python
print("Number of growth stages:", len(stage_reference))

stage_reference
```

```python
parameter_sources = (
    crop_water_clean[
        ["source_url", "data_status"]
    ]
    .drop_duplicates()
    .reset_index(drop=True)
)

parameter_sources
```

```python
reference_path = (
    "/content/Smart-Agricultural-Water-Management/"
    "weather_soil_water_modeling/data/reference/"
)

crop_calendar_clean.to_csv(
    reference_path + "crop_calendar_reference.csv",
    index=False
)

crop_water_clean.to_csv(
    reference_path + "crop_water_parameters_reference.csv",
    index=False
)

season_clean.to_csv(
    reference_path + "season_reference.csv",
    index=False
)

rice_variety_clean.to_csv(
    reference_path + "rice_variety_reference_prepared.csv",
    index=False
)

stage_reference.to_csv(
    reference_path + "rice_growth_stage_kc_reference.csv",
    index=False
)

parameter_sources.to_csv(
    reference_path + "rice_parameter_sources.csv",
    index=False
)

print("Rice reference datasets saved successfully.")
```
