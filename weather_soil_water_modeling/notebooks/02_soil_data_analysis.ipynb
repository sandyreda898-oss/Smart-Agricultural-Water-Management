```python
soil["texture_total_pct"] = (
    soil["sand_pct"] +
    soil["silt_pct"] +
    soil["clay_pct"])

soil[[
        "point_id",
        "sand_pct",
        "silt_pct",
        "clay_pct",
        "texture_total_pct"]].head(10)
```

```python
soil["texture_total_pct"].describe()
```

```python
soil[
    soil["texture_total_pct"].notna() &
    (
        (soil["texture_total_pct"] < 99) |
        (soil["texture_total_pct"] > 101))][
    [
        "point_id",
        "sand_pct",
        "silt_pct",
        "clay_pct",
        "texture_total_pct"
    ]]
```

```python
print(
    "Invalid sand values:",
    ((soil["sand_pct"] < 0) | (soil["sand_pct"] > 100)).sum()
)

print(
    "Invalid silt values:",
    ((soil["silt_pct"] < 0) | (soil["silt_pct"] > 100)).sum()
)

print(
    "Invalid clay values:",
    ((soil["clay_pct"] < 0) | (soil["clay_pct"] > 100)).sum()
)

print(
    "Invalid pH values:",
    ((soil["ph_water"] < 0) | (soil["ph_water"] > 14)).sum()
)

print(
    "Invalid depth records:",
    (soil["top_depth_cm"] >= soil["bottom_depth_cm"]).sum()
)
```

```python
soil_prepared = soil.merge(
    awc,
    on=["point_id", "smu_id"],
    how="left",
    validate="one_to_one")

print("Merged shape:", soil_prepared.shape)

soil_prepared.head()
```

```python
soil_prepared = soil_prepared.merge(
    grid,
    on="point_id",
    how="left",
    validate="one_to_one")

print("Final prepared shape:", soil_prepared.shape)

soil_prepared.head()
```

```python
soil_prepared["exclude_from_model"] = (
    soil_prepared["awc_use_recommendation"] == "EXCLUDE")

soil_prepared[
    [
        "point_id",
        "soil_data_status",
        "awc_use_recommendation",
        "review_status",
        "exclude_from_model"]]
```

```python
soil_clean = soil_prepared[
    ~soil_prepared["exclude_from_model"]].copy()

print("Prepared soil rows:", len(soil_prepared))
print("Clean soil rows:", len(soil_clean))
print("Excluded rows:", len(soil_prepared) - len(soil_clean))
```

```python
prepared_output_path = (
    "/content/Smart-Agricultural-Water-Management/"
    "weather_soil_water_modeling/"
    "data/processed/soil_prepared.csv")

soil_prepared.to_csv(
    prepared_output_path,
    index=False)

print("Prepared soil dataset saved.")
print(prepared_output_path)
```

```python
clean_output_path = (
    "/content/Smart-Agricultural-Water-Management/"
    "weather_soil_water_modeling/"
    "data/processed/soil_clean.csv")

soil_clean.to_csv(
    clean_output_path,
    index=False)

print("Clean soil dataset saved.")
print(clean_output_path)
```
